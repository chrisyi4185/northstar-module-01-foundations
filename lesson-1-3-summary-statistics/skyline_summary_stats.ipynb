| Column | Scale of measurement |
|--------|---------------------|
| enrollment_id | Nominal |
| course_name | Nominal |
| enrollment_year | Interval |
| final_grade | Ordinal |
| hours_studied | Ratio |
| completion_status | Ordinal |


```python
import pandas as pd
import numpy as np
```

```python
enrollments = pd.read_csv("../lesson-1-2-types-of-data/skyline_enrollments.csv")
enrollments.head(10)
```

```python
hours_studied_mean = enrollments["hours_studied"].mean()
print("Mean hours studied:", hours_studied_mean)

hours_studied_median = enrollments["hours_studied"].median()
print("Median hours studied:", hours_studied_median)

hours_studied_mode = enrollments["hours_studied"].mode()[0]
print("Mode hours studied:", hours_studied_mode)

hours_studied_std = enrollments["hours_studied"].std()
print("Standard deviation of hours studied:", hours_studied_std)

hours_studied_variance = enrollments["hours_studied"].var()
print("Variance of hours studied:", hours_studied_variance)

hours_studied_min = enrollments["hours_studied"].min()
print("Minimum hours studied:", hours_studied_min)

hours_studied_max = enrollments["hours_studied"].max()
print("Maximum hours studied:", hours_studied_max)

hours_studied_q1 = enrollments["hours_studied"].quantile(0.25)
print("First quartile (Q1) of hours studied:", hours_studied_q1)

hours_studied_q3 = enrollments["hours_studied"].quantile(0.75)
print("Third quartile (Q3) of hours studied:", hours_studied_q3)

hours_studied_iqr = hours_studied_q3 - hours_studied_q1
print("Interquartile range (IQR) of hours studied:", hours_studied_iqr)
```

```python
enrollment_years_mean = enrollments["enrollment_year"].mean()
print("Mean enrollment year:", enrollment_years_mean)

enrollment_years_median = enrollments["enrollment_year"].median()
print("Median enrollment year:", enrollment_years_median)

enrollment_years_mode = enrollments["enrollment_year"].mode()[0]
print("Mode enrollment year:", enrollment_years_mode)

enrollment_years_std = enrollments["enrollment_year"].std()
print("Standard deviation of enrollment year:", enrollment_years_std)

enrollment_years_q1 = enrollments["enrollment_year"].quantile(0.25)
print("First quartile (Q1) of enrollment year:", enrollment_years_q1)

enrollment_years_q3 = enrollments["enrollment_year"].quantile(0.75)
print("Third quartile (Q3) of enrollment year:", enrollment_years_q3)

enrollment_years_iqr = enrollment_years_q3 - enrollment_years_q1
print("Interquartile range (IQR) of enrollment year:", enrollment_years_iqr)
```

```python
enrollment_ids_count = enrollments["enrollment_id"].count()
print("Count of enrollment IDs:", enrollment_ids_count)

enrollment_ids_unique = enrollments["enrollment_id"].nunique()
print("Count of unique enrollment IDs:", enrollment_ids_unique)

enrollment_ids_mode = enrollments["enrollment_id"].mode()[0]
print("Mode enrollment ID:", enrollment_ids_mode)
```

```python
course_names_count = enrollments["course_name"].count()
print("Count of course names:", course_names_count)

course_names_unique = enrollments["course_name"].nunique()
print("Count of unique course names:", course_names_unique)

course_names_mode = enrollments["course_name"].mode()[0]
print("Mode course name:", course_names_mode)
```

```python
final_grades_mode = enrollments["final_grade"].mode()[0]
print("Mode final grade:", final_grades_mode)

final_grades_count = enrollments["final_grade"].count()
print("Count of final grades:", final_grades_count)
```

```python
def grade_to_value(grade):
    if grade == "A":
        return 4
    elif grade == "B":
        return 3
    elif grade == "C":
        return 2
    elif grade == "D":
        return 1
    elif grade == "F":
        return 0
    else:
        return np.nan

def value_to_grade(value):
    if value == 4:
        return "A"
    elif value == 3:
        return "B"
    elif value == 2:
        return "C"
    elif value == 1:
        return "D"
    elif value == 0:
        return "F"
    else:
        return np.nan


final_grades_numeric = enrollments["final_grade"].apply(grade_to_value)

final_grades_mean = final_grades_numeric.mean()
print("Mean final grade (numeric):", final_grades_mean)
final_grades_mean_letter = value_to_grade(round(final_grades_mean))
print("Mean final grade (letter):", final_grades_mean_letter)
```

The ordinal data in the final grade category does not take into account that there are unequal gaps between all of the grades and percentages were not given in the initial data. The initial data was presented in letter grade format and had to be converted to numerical to calculate a mean and then converted back to letter grades by rounding. While the mean and the mode appear the same in this sample, there is the potential that in other data sets the mode would be a more accurate measure than the mean.
