```python
import numpy as np
import pandas as pd
from scipy import stats
import plotly.express as px
```

```python
enrollments = pd.read_csv("../lesson-1-2-types-of-data/skyline_enrollments.csv")

n = len(enrollments)
sample_mean = enrollments["hours_studied"].mean()
sample_std = enrollments["hours_studied"].std()

print(f"Sample size: n = {n}")
print(f"Sample mean: {sample_mean}")
print(f"Sample standard deviation: {sample_std}")

```

```python
standard_error = sample_std / np.sqrt(n)
z = 1.96  # for a 95% confidence interval
margin_of_error = z * standard_error
confidence_interval = (sample_mean - margin_of_error, sample_mean + margin_of_error)
print(f"95% Confidence Interval: {confidence_interval}")
print(f"Margin of Error: {margin_of_error}")
print(f"Standard Error: {standard_error}")
```

The average hours studied in the sample is 30.6 hours. Based on the sample size of 100 students and a sample standard deviation of 15.99, a 95% confidence interval for the true mean is 27.50 to 33.77 hours. With 95% confidence, the true average falls somwhere in that range.
