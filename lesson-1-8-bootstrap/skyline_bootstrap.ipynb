# Bootstrap CI for the mean

Compute a 95% bootstrap CI for the mean of hours_studied. Use B = 10,000 iterations. Compare the bootstrap CI to the classical CI you computed in Lesson 1.5's Practice Exercise. Report:

Sample mean
Classical 95% CI
Bootstrap 95% CI
A one-sentence note on whether the two CIs are similar (you should expect them to be, for a mean on this sample size)

```python
import pandas as pd
import numpy as np

enrollments = pd.read_csv("../lesson-1-2-types-of-data/skyline_enrollments.csv")
data = enrollments["hours_studied"].values
n = len(data)
B = 10000
hours_studied = enrollments["hours_studied"].values
sample_mean = enrollments["hours_studied"].mean()
sample_std = enrollments["hours_studied"].std()

print(f"Sample size: n = {n}")
print(f"Sample mean: {sample_mean:.2f}")
print(f"Sample std: {sample_std:.2f}")

standard_error = sample_std / np.sqrt(n)
z = 1.96  # for a 95% confidence interval
margin_of_error = z * standard_error
ci_lower = sample_mean - margin_of_error
ci_upper = sample_mean + margin_of_error

print(f"Standard error: {standard_error:.2f}")
print(f"Margin of error: {margin_of_error:.2f}")
print(f"95% Confidence Interval for the mean: ({ci_lower:.2f}, {ci_upper:.2f})")

```

```python
np.random.seed(42)

boot_means = np.array([
    np.random.choice(data, size=n, replace=True).mean()
    for _ in range(B)
])

boot_medians = np.array([
    np.median(np.random.choice(data, size=n, replace=True))
    for _ in range(B)
])
boot_p95 = np.array([
    np.percentile(np.random.choice(data, size=n, replace=True), 95)
    for _ in range(B)
])

ci_p95 = np.percentile(boot_p95, [2.5, 97.5])
ci_mean = np.percentile(boot_means, [2.5, 97.5])
ci_median = np.percentile(boot_medians, [2.5, 97.5])

print(f"Sample Size: n = {n}")
print(f"Sample Mean: {sample_mean:.2f}")
print(f"Sample Median: {np.median(data):.2f}")
print(f"Sample Standard Deviation: {sample_std:.2f}")
print(f"95% Bootstrap Confidence Interval for the mean: ({ci_mean[0]:.2f}, {ci_mean[1]:.2f})")
print(f"95% Bootstrap Confidence Interval for the median: ({ci_median[0]:.2f}, {ci_median[1]:.2f})")
print(f"95% Bootstrap Confidence Interval for the 95th percentile: ({ci_p95[0]:.2f}, {ci_p95[1]:.2f})")
```

```python
import plotly.express as px

bootstrap_df = pd.DataFrame({
    "boot_mean": boot_means,
    "boot_median": boot_medians
})

fig_mean = px.histogram(
    bootstrap_df,
    x="boot_mean",
    nbins=50,
    title="Bootstrap Distribution of the Mean: Hours Studied"
)
fig_mean.add_vline(x=ci_mean[0], line_dash="dash", annotation_text="2.5%")
fig_mean.add_vline(x=ci_mean[1], line_dash="dash", annotation_text="97.5%")
fig_mean.show()

fig_median = px.histogram(
    bootstrap_df,
    x="boot_median",
    nbins=50,
    title="Bootstrap Distribution of the Median: Hours Studied"
)
fig_median.add_vline(x=ci_median[0], line_dash="dash", annotation_text="2.5%")
fig_median.add_vline(x=ci_median[1], line_dash="dash", annotation_text="97.5%")
fig_median.show()
```

The CIs are indeed similar between classical and bootstrap ( 0.01 difference ).
